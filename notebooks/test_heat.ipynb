```r
# 1. Strip the version numbers (anything after the dot) from your rownames
rownames(gene_tpm) <- sub("\\..*", "", rownames(gene_tpm))

# 2. Now proceed with the mapping as before
ensembl_ids <- rownames(gene_tpm)
gene_info <- bitr(ensembl_ids, fromType="ENSEMBL", toType="SYMBOL", OrgDb=org.Mm.eg.db)

# 3. Continue with the rest of your script
marker_genes_high <- c("Fos", "Il1a", "Il18r1", "Nos2", "Il23a", "Tmem40", "Myo1b", "Notch4")
marker_genes_low <- c("Cdx2", "Hoxa13")

high_ids <- gene_info$ENSEMBL[gene_info$SYMBOL %in% marker_genes_high]
low_ids <- gene_info$ENSEMBL[gene_info$SYMBOL %in% marker_genes_low]

heatmap_matrix <- gene_tpm[c(high_ids, low_ids), ]
rownames(heatmap_matrix) <- gene_info$SYMBOL[match(rownames(heatmap_matrix), gene_info$ENSEMBL)]

# 4. Draw the heat map
plot_data <- log2(heatmap_matrix + 1)
Heatmap(as.matrix(plot_data), 
        name = "Log2(TPM+1)",
        column_title = "Subclones (Met-High vs Met-Low)",
        show_column_names = TRUE,
        cluster_rows = FALSE,
        row_split = rep(c("Met-High Markers", "Met-Low Markers"), c(length(high_ids), length(low_ids))),
        col = colorRampPalette(c("blue", "white", "red"))(100))
```
